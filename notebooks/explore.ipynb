# 564 genres, two dials
Load the catalogue, find a style, and see where it sits on the Weirdness × Style Influence map.

```python
import pandas as pd
df = pd.read_parquet('https://raw.githubusercontent.com/samuelgrupolimex-prog/suno-style-recipes/main/styles.parquet')
df.shape
```

```python
# Find a style by name (case-insensitive)
query = 'bolero'
df[df['style'].str.contains(query, case=False)][['style','bpm','weirdness','style_influence','style_prompt']]
```

```python
# The styles closest to a setting you like
w, si = 30, 92
df.assign(dist=((df.weirdness-w)**2 + (df.style_influence-si)**2)**0.5).nsmallest(10,'dist')[['style','family','weirdness','style_influence']]
```

```python
import matplotlib.pyplot as plt
ax = df.plot.scatter(x='weirdness', y='style_influence', alpha=.5, figsize=(10,5))
ax.set_title('564 genres. Two dials.')
ax.set_ylim(76, 101); plt.show()
```

```python
# Style Influence by family: how tight each family holds its genre
df.groupby('family').style_influence.agg(['min','median','max']).sort_values('median')
```
